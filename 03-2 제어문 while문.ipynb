### 반복문 while
* while 반복문은 `조건식으로만 동작`하고 반복할 코드안에는 조건식에 영향을 주는 변화식을 정의한다.

> `while 조건식`:<br>
    실행문1
    ....
    실행무n

'열 번 찍어 안 넘어가는 나무 없다'라는 속담을 파이썬 프로그램으로 만들면 다음과 같다.

```python
treeHit = 0
while treeHit < 10:
    treeHit = treeHit +1
    print(f"나무를 {treeHit}번 찍었습니다.")
    if treeHit == 10:
        print("나무 넘어갑니다.")
```

### while 문 만들기

이번에는 여러 가지 선택지 중 하나를 선택해서 입력받는 예제

### <mark>주의사항 --> input() 또는 while 문 + input() 또는 대화명 프로그램은 .py (프로그램)으로 실행하세요</mark>

```python
prompt = """
1. Add
2. Del
3. List
4. Quit

Enter number: """
number = 0
while number != 4:
    print(prompt)
    number = int(input())
```

### while 문 강제로 빠져나가기 --> if문 + break

커피 자판기를 예로 들어 보자. 만약 커피가 떨어졌다면 판매를 중단하고 '판매 중지' 문구를 보여 주어야 한다.

```python
coffee = 10
money = 300
while money:
    print("돈을 받았으니 커피를 줍니다.")
    coffee = coffee -1
    print(f"남은 커피의 양은 {coffee}개입니다.")
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        break
```

### while 문의 맨 처음으로 돌아가기 --> continue

1부터 10까지의 숫자 중에서 홀수만 출력하는 것을 while 문을 사용해서 작성해 보자.

```python
a = 0
while a < 10:
    a = a + 1
    if a % 2 == 0: continue
    print(a)
```

### 무한 루프

다음은 무한 루프의 예이다.

* 쥬피터 노트북에서는 쥬피터 노트북 서버를 중지해야만 합니다.  Ctrl + c --> IDLE 경우,  우리는 인텔리제이 --> Ctrl + c 는 복사 명령임

```python
while True:
    print("Ctrl+C를 눌러야 while문을 빠져나갈 수 있습니다.")
```

```python
count = 0
while count < 3:
    print(f"카운트: {count}")
    count += 1
else:
    print("while 문이 정상 종료되었습니다.")
```

만약 위 예제에서 break로 빠져나가면 else 절의 문장은 출력되지 않는다.

```python
count = 0
while count < 3:
    if count == 2:
        break
    print(f"카운트: {count}")
    count += 1
else:
    print("while 문이 정상 종료되었습니다.")
```

### 중첩된 while 문

while 문 안에 또 다른 while 문을 사용할 수 있다. 이를 중첩된 while 문이라고 한다.

다음은 중첩된 while 문으로 구구단 2단과 3단을 출력하는 예이다.

```python
i = 2
while i <= 3:
    j = 1
    while j <= 9:
        print(f"{i} x {j} = {i*j}")
        j += 1
    i += 1
```

* 바깥쪽 while 문은 단(i)을 반복하고, 안쪽 while 문은 각 단의 곱셈(j)을 반복한다.
* 이처럼 중첩된 while 문은 바깥쪽 반복 한 번마다 안쪽 반복을 처음부터 끝까지 수행한다.

중첩된 while 문에서 break나 continue를 사용하면 가장 가까운 while 문에만 영향을 준다.
