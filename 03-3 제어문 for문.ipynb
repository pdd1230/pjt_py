range 함수의 예시 살펴보기

리스트 컴프리헨션 (내포) 사용하기

일반적인 for문

```python
a = [1,2,3,4]
result = []
for num in a:
    result.append(num*3)

print(result)
```

리스트 컴프리헨션을 사용하면 다음과 같이 간단하게 작성할 수 있다.

```python
a = [1,2,3,4]
result = [num * 3 for num in a]
print(result)
```

만약 [1, 2, 3, 4] 중에서 짝수에만 3을 곱하여 담고 싶다면 리스트 컴프리헨션 안에 if 조건문을 사용하면 된다.

```python
a = [1,2,3,4]
result = [num * 3 for num in a if num % 2 == 0]
print(result)
```

리스트 컴프리헨션의 문법은 다음과 같다. if 조건문 부분은 앞의 예제에서 볼 수 있듯이 생략할 수 있다.

### [표현식 for 항목 in 반복_가능_객체 if 조건문]

만약 구구단의 모든 결과를 리스트에 담고 싶다면 리스트 컴프리헨션을 사용하여 다음과 같이 간단하게 구현할 수 있다.

```python
result = [x*y for x in range(2,10)
          for y in range(1,10)]
print(result)
```

for 문과 break 문

> break 문은 for 문을 강제로 빠져나가고 싶을 때 사용한다.

```python
for i in range(10):
    if i == 5:
        break   #  i가 5가 되면 break 문이 실행되어 for 문을 빠져나간다.
    print(i)
```

for-else 문

> for 문 뒤에 else를 붙일 수 있다.
> for 문이 끝까지 수행되면 else 절이 실행되고, break로 빠져나가면 else 절은 실행되지 않는다.


```python
for i in range(5):
    print(i)
else:
    print("for 문이 정상 종료되었습니다.")
```

만약 break로 빠져나가면 else 절의 문장은 출력되지 않는다.

```python
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("for 문이 정상 종료되었습니다.")
```

for-else 문은 자주 사용하지는 않지만, 리스트에서 원하는 값을 찾았는지 여부를 판단할 때 유용하다. break 없이 끝났다면 "찾지 못했다"는 의미로 else 절을 활용할 수 있다.

enumerate 함수 활용하기

리스트의 순서(인덱스)와 값을 함께 구하고 싶을 때는 enumerate 함수를 사용하면 편리하다.

```python
fruits = ['apple', 'banana', 'orange']
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")
```

enumerate는 0부터 시작하는 인덱스 번호를 자동으로 생성해 준다. 시작 번호를 변경하고 싶다면 다음과 같이 사용할 수 있다.

```python
fruits = ['apple', 'banana', 'orange']
for i, fruit in enumerate(fruits, 1):  # 1부터 시작
    print(f"{i}: {fruit}")
```

zip 함수로 여러 리스트 함께 순회하기

> 두 개 이상의 리스트를 동시에 순회하고 싶을 때는 zip 함수를 사용한다.

```python
names = ['홍길동', '김철수', '이영희']
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"{name}: {score}점")
```

세 개 이상의 리스트도 함께 순회할 수 있다.

```python
names = ['홍길동', '김철수', '이영희']
korean = [85, 92, 78]
english = [90, 88, 95]
for name, kor, eng in zip(names, korean, english):
    print(f"{name}: 국어 {kor}점, 영어 {eng}점")
```

```python
# 문제1. 예상 숫자를 맞추기 게임 --> # number_binggo.py

import random

print('>> 숫자 맞히기 게임 <<')

com = random.randint(1, 10) # 1 ~ 10 랜덤 생성 ?

while True:
    my = int(input('예상 숫자를 입력하세요 : '))
    if com == my:
        print(f'예상 숫자를 맞추었습니다.')
        break
    elif my>com:
        print(f'예상숫자한 숫자보다 작습니다.')
    else:
        print(f'예상숫자 숫자보다 큽니다.')
```

```python
# 문제3. 별표 출력(2)

'''
    *
   ***
  *****
 *******
*********

'''
```

```python
for i in range(1,10,2):
    print(f'{"*"*i:^9}')
```

* range(1, 10, 2): 1부터 시작해서 10이 되기 전까지 2씩 증가하는 숫자를 만듭니다
> 즉, 변수 i에는 1, 3, 5, 7, 9가 차례대로 들어갑니다. (총 5번 반복)
* "*" * i: 파이썬은 문자열에 숫자를 곱할 수 있습니다.
> i가 3일 때는 ***, 5일 때는 *****가 됩니다.
* :^9 (핵심): f-string(포맷 스트링)의 정렬 기능입니다.
> 총 9칸의 빈 공간을 마련해 두고, 그 안에서 텍스트를 가운데 정렬(^) 하라는 뜻입니다.

```python
for i in range(1,10,2):
    print(f'{"*"*i:^9}')
for i in range(7,0,-2):
    print(f'{"*"*i:^9}')
```
