````python
# A. numpy

# 1. 난수발생
# print(dir(np.random))

data = np.random.rand()
print(data, type(data))

data = np.random.rand(50)
print(data, type(data))
````

````python
# B. pandas
# 1. 1차원 자료 : Series

data = np.random.rand(50)

s = pd.Series(data)
print(s, type(s))
````

````python
%matplotlib inline
# 시각화(pandas 고유기능으로 시각화) - ndarray로 시각화

s.plot()
````

````python
# 2. 2차원 자료 : DataFrame

df = pd.DataFrame(data)
print(df, type(df))

display(df, type(df))
````

````python
df.plot()
````

````python
# numpy로 랜덤하게 10*3열의 2차원배열(행렬 만들기)

data = np.random.rand(10, 3)
print(data, type(data))
````

````python
df = pd.DataFrame(data)
display(df, type(df))
````

````python
df = pd.DataFrame(data, columns=['2018','2019','2020'])
display(df)
````

````python
# 2차원 자료로 시각화
df.plot()
````

````python
# bar
df.plot(kind = 'bar')
````

````python
# barh

df.plot(kind='barh')
````

````python
# area

df.plot(kind = 'area')
df.plot(kind = 'area', stacked = False)
````

````python
# pie
s = pd.Series(data=np.random.rand(5), index = [2015,2016,2017,2018,2019])
s.plot(kind = 'pie', fontsize=11, title = 'sales test', figsize=(10,5), autopct = "%.2f%%")
````
